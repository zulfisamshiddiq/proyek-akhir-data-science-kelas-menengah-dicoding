## **Cleaning Data**

### **Membersihkan customer_df**

#### Menghilangkan duplikasi data

```python
customers_df.drop_duplicates(inplace=True)
print(f"Jumlah duplikasi : {customers_df.duplicated().sum()}")
```

#### Menangani missing value

```python
# Menampilkan rows data missing value dari column gender
customers_df[customers_df.gender.isna()]
```

Terlalu banyak informasi penting untuk dihapus. Maka gunakan metode *imputation* (ganti dengan mean/median/mode)

```python
# Identifikasi nilai dominan/mode pada column gender
customers_df.gender.value_counts()
```

```python
# Gunakan 725 untuk mengisi missing value
customers_df.fillna(value="Prefer not to say", inplace=True)
print(f"Jumlah missing value :\n\n{customers_df.isna().sum()}")
```

#### Menangani *inaccurate value*

```python
# Menampilkan rows inaccurate value pada column age
customers_df[customers_df.age == customers_df.age.max()]
```

Angka 700 terjadi karena *humman error* harusnya 70 bukan 700. Cara mengatasinya pakai methode **replace()**

```python
# Atasi inaccurate value pada column age
customers_df.age.replace(customers_df.age.max(), 70, inplace=True)
customers_df[customers_df.age == customers_df.age.max()]
```

700 sudah terganti dengan 70, namun setelah di cek lagi masih ada *humman error* lainnya, harusnya 50 ditulis 500

```python
customers_df.age.replace(customers_df.age.max(), 50, inplace=True)
customers_df.describe()
```

## **EDA order_df**

Selanjutnya untuk mengetahui berapa lama dari order sampai delivery? Berdasarkan informasi data order (order_id, customer_id, order_date, dan delivery_date), kita dapat membuat new column untuk memuat waktu pengiriman tiap order dengan menghitung selisih **delivery_date** dan **order_date** dan disimpan dalam new column **delivery_time**.

Gunakan methode **apply()** untuk operasi terhadap setiap elemn dalam sebuah column DF atau series (1D). Operasinya adalah menghitung jumlah detik dari **delivery_time** menggunakan method **total_seconds()** dengan mengubah satunnya menjadi hari (dibagi 86400) dan ambil bilangan bulatnya. 1 hari = 86.300 detik

```python
delivery_time = orders_df["delivery_date"] - orders_df["order_date"]
# delivery - order

delivery_time = delivery_time.apply(lambda x: x.total_seconds())
# apply()   : menerapkan fungsi ke setiap rows data
# lambda x  : fungsi anonim (fungsi singkat tanpa nama)
#x.total_senconds() : ubah selisih waktu jadi total detik

orders_df["delivery_time"] = round(delivery_time / 86400)
# round ()          : membulatkan jumalh hari ke bilangan terdekat
# delivery/86400    : mengubah detik menjadi perhari

orders_df.head()
```

```python
# Rangkuman statistik
orders_df.describe(include="all")
```

Dari informasi diatas waktu pengiriman mulai dari 1-27 hari dengan rata-rata 14 hari. Selanjutnya menggabungkan data orders_df dengan customers_df

## **Data Visualization**

**1. Bagaimana Performa Penjualan dan Revenue Perusahaan dalam Beberapa Bulan Terakhir?**

Gunakan pandas dengan method **resample()** untuk mengubah frekuensi atau resampling terhadap DF yang memiliki komponen *time series* dengan mendefinisikan parameter **rule** (mengatur target konservasi) dan **on** (nama kolom bertipe datatime yang akan diubah frekuensinya).

```python
monthly_orders_df = all_df.resample(rule="M", on="order_date").agg({
    "order_id" : "nunique",
    "total_price" : "sum"
})

monthly_orders_df.head()
```

Supaya mudah dalam melihat data, alangkah baiknya dibuatkan visualisasi menggunakan line chart

```python
monthly_orders_df = all_df.resample(rule='M', on='order_date').agg({
    "order_id" : "nunique",
    "total_price" : "sum"
})
monthly_orders_df.index = monthly_orders_df.index.strftime("%B")

monthly_orders_df = monthly_orders_df.reset_index()
monthly_orders_df.rename(columns={
    "order_id" : "order_count",
    "total_price" : "revenue"
}, inplace=True)

plt.figure(figsize=(10, 5))
plt.plot(monthly_orders_df["order_date"], monthly_orders_df["order_count"], marker="o", linewidth=2, color="#72BCD4")
plt.title("Number of orders per Month (2021)", loc="center", fontsize=20)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

Jumlah order terbanyak terjadi pada bulan maret, dan penurunan signifikan pada bulan Februaru, April, Mei, dan Oktober.

Hal ini akan berdampak pada total *revenue* yang diperoleh perusahaan.

```python
plt.figure(figsize=(10, 5))
plt.plot(
    monthly_orders_df["order_date"],
    monthly_orders_df["revenue"],
    marker='o', 
    linewidth=2,
    color="#72BCD4"
)
plt.title("Total Revenue per Month (2021)", loc="center", fontsize=20)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

Sama dengan keadaan order, revenue memiliki penurunan yang signifikan pada bulan Februrari, April, Mei, dan Oktober. Kita harus mencari tahu penyebab terjadinya penurunan tersebut dengan mempertimbangkan banyak hal seperti keberadaan kompetitor, campign. Namun untuk setudi kausus ini tidak memiliki cukup informasi terkait hal tersebut

### **RFM Analysis**
Untuk menjawab pertanyaan terakhir analisis terakhir, bisa menggunakan analisis lanjutan ini untuk melakukan segmentasi pelanggan seperti pengelompokkan pelanggan ke dalam beberapa kategori berdasarkan 3 kategori : <br>
- **Recency :** untuk melihat kapan pelanggan terakhir kali melakukan transaksi
- **Frequency :** untuk mengidentifikasi seberapa sering pelanggan melakukan transaksi
- **Monetary :** untuk mengidentifikasi seberapa besar revenue yang berasal dari pelanggan tertentu

Dengan hal ini kita bisa mengidentifikasi pelanggan mana yang memiliki *high value* (sering melakukan transaksi dan menghasilkan revenue yang besar) dan low value.

Melakukan RFM bisa dengan cara membuat DF untuk menampung informasi terkait tidak parameter tersebut

```python
rfm_df = all_df.groupby(by="customer_id", as_index=False).agg({
    "order_date": "max", # mengambil tanggal order terakhir
    "order_id": "nunique", # menghitung jumlah order
    "total_price": "sum" # menghitung jumlah revenue yang dihasilkan
})
rfm_df
```

```python
rfm_df = all_df.groupby(by="customer_id", as_index=False).agg({
    "order_date": "max", # mengambil tanggal order terakhir
    "order_id": "nunique", # menghitung jumlah order
    "total_price": "sum" # menghitung jumlah revenue yang dihasilkan
})
rfm_df.columns = ["customer_id", "max_order_timestamp", "frequency", "monetary"]
rfm_df
```

```python
rfm_df = all_df.groupby(by="customer_id", as_index=False).agg({
    "order_date": "max", # mengambil tanggal order terakhir
    "order_id": "nunique", # menghitung jumlah order
    "total_price": "sum" # menghitung jumlah revenue yang dihasilkan
})
rfm_df.columns = ["customer_id", "max_order_timestamp", "frequency", "monetary"]
 
# menghitung kapan terakhir pelanggan melakukan transaksi (hari)
rfm_df["max_order_timestamp"] = rfm_df["max_order_timestamp"].dt.date
recent_date = orders_df["order_date"].dt.date.max()
rfm_df["recency"] = rfm_df["max_order_timestamp"].apply(lambda x: (recent_date - x).days)

rfm_df.drop("max_order_timestamp", axis=1, inplace=True)
rfm_df.head()
```

Dengan data tersebut kita akan melakukan identifikasi **best customer** berdasarkan parameter RFM

Tahapan berikutnya adalah menyiapkan DF yang digunakan untuk membuat visualisasi data. Namun kita perlu membuat beberapa *helper function* seperti berikut.<br>
- **create_daily_orders_(df) :** untuk menyiapkan daily_orders_df
- **create_sum_order_items_df(df) :** untuk menyiapkan sum_orders_items_df
- **create_sum_order_items_df(df) :** untuk menyiapkan sum_order_items_df
- **create_bygender_df(df) :** untuk bygender_df
- **create_byage_df(df) :** untuk byage_df
- **create_bystate_df(df) :** untuk bystate_df
- **create_rfm_df(df) :** untuk rfm_df

```python
def create_daily_orders_df(df):
    daily_orders_df = df.resample(rule="D", on="order_date").agg({
        "order_id" : "nunique",
        "total_price" : "sum"
    })
    daily_orders_df = daily_orders_df.reset_index()
    daily_orders_df.rename(columns={
        "order_id" : "order_count",
        "total_price" : "revenue"
    }, inplace=True)
    
    return daily_orders_df
```

```python
def create_sum_order_items_df(df):
    sum_order_items_df = df.groupby("product_name").quantity_x.sum().sort_values(ascending=False).reset_index()
    return sum_order_items_df
```

```python
def create_bygender_df(df):
    bygender_df = df.groupby(by="gender").customer_id.nunique().reset_indec()
    bygender_df.rename(columns={
        "customer_id" : "customer_count"
    }, inplace=True)
    
    return bygender_df
```

```python
def creat_byage_df(df):
    byage_df = df.groupby(by="age_group").customer_id.nunique().reset_index()
    byage_df.rename(columns={
        "customer_id" : "customer_count"
    }, inplace=True)
    byage_df["age_group"] = pd.Categorical(byage_df["age_group"], ["Youth", "Adults", "SSeniors"])
    
    return byage_df
```

```python
def create_bystate_df(df):
    bystate_df = df.groupby(by="state").customer_id.nunique().reset_index()
    bystate_df.rename(columns={
        "customer_id" : "customer_count"
    }, inplace=True)
    
    return bystate_df
```

```python
def create_rfm_df(df):
    rfm_df = df.groupby(by="customer_id", as_index=False).agg({
        "order_date": "max", #mengambil tanggal order terakhir
        "order_id": "nunique",
        "total_price": "sum"
    })
    rfm_df.columns = ["customer_id", "max_order_timestamp", "frequency", "monetary"]
    
    rfm_df["max_order_timestamp"] = rfm_df["max_order_timestamp"].dt.date
    recent_date = df["order_date"].dt.date.max()
    rfm_df["recency"] = rfm_df["max_order_timestamp"].apply(lambda x: (recent_date - x).days)
    rfm_df.drop("max_order_timestamp", axis=1, inplace=True)
    
    return rfm_df
```
